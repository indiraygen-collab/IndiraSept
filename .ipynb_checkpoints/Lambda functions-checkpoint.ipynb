**lamda functions**
- it is a function in a single line
- many arguments , but only one expression
-     - lambda <arguments>:<expression>

```python
def add(a):
    print(a+10)
add(100)
```

```python
add=lambda a : a+10
add(20)
```

```python
#def cub(x):
#    print(x**3)
cube=lambda x:x**3
cube(9)
```

```python
#def avg(a,b,c):
#print((a+b+c)/3)
avg= lambda a,b,c:(a+b+c)/3
avg(6,7,8)
```

```python
#def avg1(a,b,c=70):
 #   print((a+b+c)/3)
avg1= lambda a,b,c=50:round((a+b+c)/3)
avg1(6,7)
```

**if - else**

```python
def even_odd(a):
    if a%2==0:
        print(f'{a} is even')
    else:
        print(f'{a} is odd')
even_odd(9)
    
```

```python
even_odd=lambda a:print(f'{a} is even') if a%2==0 else print(f'{a} is odd')
even_odd(9)
```

```python
def max (a,b):
    if a>b:
        print(a)
    else:
        print(b)
        
```

```python
max1=lambda a,b: a if a>b else b
max1(8,9)
```

```python
l=[1,2,3,4,5,6]

list(map(lambda i: f'{i} is even' if i%2==0 else f'{i} is odd' ,l))
```

```python
l=[1,2,3,4,5,6]
list(filter(lambda i: i%2==0,l))
```
